```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
from pathlib import Path
```

```python
turn_folder = Path("../pilot_llm_evaluation_turn/csv/")
holistic_folder = Path("../pilot_llm_evaluation_holistic/csv/")
```

## Make Turn Dataset 

```python
# Make Turn Dataset


# find all csv files that contain 'ALL' in the filename
all_files = list(turn_folder.glob("*ALL*.csv"))

print(f"Found {len(all_files)} files:")
for f in all_files:
    print(" -", f.name)

# read and combine
data = [pd.read_csv(f) for f in all_files]
turn_df = pd.concat(data, ignore_index=True)

print("Combined shape:", turn_df.shape)
```

```python
turn_df.head()
```

```python
# Map scenario codes to condition names (group_id)
code_to_group = {
    "L1_frustrated_multi": "Group A",
    "L5_neutral_multi": "Group A",
    "L10_angry_multi": "Group A",

    "L1_neutral_multi": "Group B",
    "L5_angry_multi": "Group B",
    "L10_neutral_multi": "Group B",

    "L1_angry_multi": "Group C",
    "L5_frustrated_multi": "Group C",
    "L10_frustrated_multi": "Group C",
}

turn_df.insert(2,'group_id', 2)
turn_df['group_id'] = turn_df['scenario_id'].map(code_to_group)
turn_df['group_id'].value_counts(dropna=False)
```

## Make Holistic Dataset 

```python
# Make Turn Dataset


# find all csv files that contain 'ALL' in the filename
all_files = list(holistic_folder.glob("*ALL*.csv"))

print(f"Found {len(all_files)} files:")
for f in all_files:
    print(" -", f.name)

# read and combine
holistic = [pd.read_csv(f) for f in all_files]
hol_df = pd.concat(holistic, ignore_index=True)

print("Combined shape:", hol_df.shape)
```

```python
hol_df.insert(2,'group_id', 2)
hol_df['group_id'] = hol_df['scenario_id'].map(code_to_group)
hol_df['group_id'].value_counts(dropna=False)
```

```python
hol_df.head()
```

# Save as CSV

```python
hol_df.to_csv('llm_holistic.csv', index=False)
turn_df.to_csv('llm_turn.csv', index=False)
```
